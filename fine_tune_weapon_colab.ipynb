# Fine-Tune Weapon Detection YOLOv8 Model (GPU Accelerated)
This notebook fine-tunes `threat-yolov8n.pt` on a weapon detection dataset using free Google Colab GPU (T4).

```python
# 1. Install dependencies
!pip install -q ultralytics huggingface_hub
```

```python
# 2. Download the base threat model
from huggingface_hub import hf_hub_download

model_path = hf_hub_download(
    repo_id="Subh775/Threat-Detection-YOLOv8n",
    filename="weights/best.pt"
)
print("Base model downloaded to:", model_path)
```

```python
# 3. Clone / download training dataset
!git clone https://huggingface.co/datasets/Simuletic/cctv-weapon-dataset dataset

# Create dataset.yaml
yaml_content = """
path: /content/dataset/samples
train: images
val: images
names:
  0: Gun
  1: Explosion
  2: Grenade
  3: Knife
"""
with open("data.yaml", "w") as f:
    f.write(yaml_content.strip())
print("data.yaml written successfully.")
```

```python
# 4. Run fine-tuning on GPU
from ultralytics import YOLO

model = YOLO(model_path)
results = model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,  # GPU
    project="weapon_runs",
    name="finetuned"
)
```

```python
# 5. Download your trained weights to put into newcctv/models/weapon/threat-yolov8n.pt
from google.colab import files
files.download("weapon_runs/finetuned/weights/best.pt")
```
